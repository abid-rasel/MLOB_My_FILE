```python
py_array=[1,4,5,"Hello","Abid"]
np_array=np.array([1,2,3,4,5,6],dtype=np.int8)
```

```python
print(sys.getsizeof(py_array))  # size of the list container only, not the elements size

print(sys.getsizeof(np_array))  # returns the size of the NumPy array object, not just the raw data.
print(np_array.nbytes)  # Actual np.array size
print(np_array.itemsize)
```

```python
import sys
import numpy as np

py_arr = [1, 2, 3, 4, 5]

np_array = np.array([1, 2, 3, 4, 5], dtype=np.int8)

print("Python list:", sys.getsizeof(py_arr))
print("NumPy array:", sys.getsizeof(np_array))
print("NumPy data:", np_array.nbytes)
```

```python
# size of py_list
total_size = sys.getsizeof(py_array) # size of the list container only, not the elements size

for item in py_array:
    total_size += sys.getsizeof(item)

print(total_size)
```

```python
import sys

py_arr = [1, 2, 3, 4, 5]

list_memory = sys.getsizeof(py_arr)
items_memory = sum(sys.getsizeof(item) for item in py_arr)

total_memory = list_memory + items_memory

print("List container:", list_memory)
print("Items:", items_memory)
print("Total:", total_memory)
```

```python
import time
import numpy as np
size=10000000
py_list=list(range(size))
np_arr=np.array(range(size))
```

```python
start_time=time.time()
py_result=[i*2 for i in py_list]
py_time=time.time()-start_time
print(py_time)
```

```python
start_time=time.time()
np_result=np_arr*2
np_time=time.time()-start_time
print(np_time)
```

```python
print(f"numpy array is {py_time/np_time:.2f} times faster than python list")
```
