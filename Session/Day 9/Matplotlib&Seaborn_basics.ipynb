```python
""" alpha= 0      Completely invisible

0.2    Very transparent

0.5    Half transparent

0.8    Almost solid

1.0    Completely visible"""
```

```python
import matplotlib.pyplot as plt

height = [160,165,170,175,180]
weight = [50,55,63,70,76]

plt.figure(figsize=(8,5))

plt.scatter(
    height,
    weight,
    color="green",
    s=100, # s = Size of the dot, instead of marker size, it is the area of the dot
    alpha=0.7,
    marker="o",
    label="Students"
)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]

boys = [55, 60, 65, 70, 75]
girls = [50, 54, 60, 66, 72]

# Create a figure
plt.figure(figsize=(8, 5))

# Scatter plot for boys
plt.scatter(
    x,
    boys,
    color="blue",
    marker="o",
    s=100,
    label="Boys"
)

# Scatter plot for girls
plt.scatter(
    x,
    girls,
    color="red",
    marker="s",
    s=100,
    label="Girls"
)

# Customize axes
plt.xlabel("Student Number")
plt.ylabel("Weight (kg)")
plt.title("Weight Comparison of Boys and Girls")

# Add grid
plt.grid(True)

# Show legend
plt.legend()

# Display the graph
plt.show()
```

```python
# Grouped bar chart
import numpy as np
import matplotlib.pyplot as plt

subjects = ["Math", "English", "Science"]

boys = [80,75,90]
girls = [78,82,88]

x = np.arange(len(subjects))

width = 0.35

plt.figure(figsize=(8,5))
plt.bar(
    x - width/2,
    boys,
    width=width,
    label="Boys"
)

plt.bar(
    x + width/2,
    girls,
    width=width,
    label="Girls"
)

plt.xticks(x, subjects)

plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.title("Scores by Subject")
plt.legend()

plt.show()
```

```python

import matplotlib.pyplot as plt

# Data
subjects = ["Math", "Physics", "Chemistry"]

boys = [85, 78, 90]
girls = [88, 82, 87]

# Create x-axis positions
x = np.arange(len(subjects))
# Output: [0, 1, 2]

# Width of each bar
width = 0.35

# Create figure
plt.figure(figsize=(9, 5))

# Boys' bars (shifted left)
plt.bar(
    x - width/2,
    boys,
    width=width,
    color="blue",
    label="Boys"
)

# Girls' bars (shifted right)
plt.bar(
    x + width/2,
    girls,
    width=width,
    color="red",
    label="Girls"
)

# Replace 0,1,2 with subject names
plt.xticks(x, subjects)

# Labels
plt.xlabel("Subjects", fontsize=12)
plt.ylabel("Scores", fontsize=12)

# Title
plt.title("Comparison of Boys' and Girls' Scores by Subject", fontsize=14)

# Grid (optional but makes the chart look professional)
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Legend
plt.legend(title="Students")

# Prevent overlapping of labels
plt.tight_layout()

# Display the chart
plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Abid","Sara","John","Mina","David",
             "Emma","Alex","Sophia","Liam","Noah"],
    "Study_Hours": [10,15,8,12,18,20,7,16,14,11],
    "Marks": [85,92,78,88,95,97,72,90,84,80],
    "Attendance": [90,95,82,88,98,99,80,93,89,85]
})
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.show()
```

This is the Most Important Seaborn Pattern

You'll see it again and again:
sns.someplot(
    data=df,
    x="column1",
    y="column2"
)
Think of it like answering three questions:

Which table contains the data?
data=df
Which column goes on the x-axis?
x="Study_Hours"
Which column goes on the y-axis?
y="Marks"

Unlike Matplotlib, you don't usually pass arrays like df["Marks"]. Instead, you give Seaborn the whole DataFrame and tell it which columns to use by name.

Seaborn's Design Philosophy

Seaborn thinks in terms of DataFrames and variables, whereas Matplotlib thinks in terms of arrays and coordinates.

That's why you'll repeatedly see the same argument names:

data=
x=
y=
(later you'll also learn hue=, style=, size=, etc.)

Once you master this pattern, learning the rest of Seaborn becomes much easier.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(8, 5))

# Line plot
sns.lineplot(data=df, x='Study_Hours', y='Marks', marker='o')

# Title and labels
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

this is one of the most valuable lessons in Seaborn because these plots help you answer:

How are the values spread?
Are there any unusual values (outliers)?
Which group has more variation?
What does the distribution look like?

These plots are used extensively in Exploratory Data Analysis (EDA) and before training machine learning models.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Abid","Sara","John","Mina","David",
             "Emma","Alex","Sophia","Liam","Noah"],
    "Study_Hours": [10,15,8,12,18,20,7,16,14,11],
    "Marks": [85,92,78,88,95,97,72,90,84,80],
    "Gender": ["Male","Female","Male","Female","Male",
               "Female","Male","Female","Male","Male"]
})
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    y="Marks",
    palette="Set3"
)

plt.title("Box Plot of Marks")

plt.show()
```

A histogram tells us

How many values are in each range.

But suppose I ask:

Where is the middle?

Where are the highest and lowest values?

Are there any unusual values?

A histogram doesn't answer these questions very clearly.

That's why we use a Box Plot.

```python
# Now you'll see two boxes, One for males.One for females. This lets you compare: Median, Spread, Outliers
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Marks",
    palette="Set2"
)

plt.show()
```

```python
# Violin Plot: A violin plot combines Box Plot, KDE Plot
plt.figure(figsize=(7,5))

sns.violinplot(
    data=df,
    x="Gender",
    y="Marks",
    palette="Set2",
    inner="quartile"
    # Show quartiles inside the violin
)

plt.show()
# A violin plot also tells us, where the data is concentrated., Wide part ↓ Many observations. Narrow part ↓, Few observations.
```

```python
# Suppose you want to see every individual student.
plt.figure(figsize=(7,5))

sns.stripplot(
    data=df,
    x="Gender",
    y="Marks",
    palette="Set2",
    jitter=True  # Add some random noise to the points for better visibility
)

plt.show()
```

```python
# Swarm Plot is like Strip Plot,but it never lets points overlap.It carefully arranges the dots.

# Think of it as, A smarter strip plot.
plt.figure(figsize=(7,5))

sns.swarmplot(
    data=df,
    x="Gender",
    y="Marks"
)

plt.show()
```

```python
# Overlaying Plots ⭐, One of Seaborn's most useful tricks. Now you see

# the summary (box)
# every individual point (strip)

# This is common in research papers.
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Marks"
)

sns.stripplot(
    data=df,
    x="Gender",
    y="Marks",
    color="black"
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name":["Abid","Sara","John","Mina","David",
            "Emma","Alex","Sophia","Liam","Noah"],
    "Marks":[85,92,78,88,95,97,72,90,84,80],
    "Gender":["Male","Female","Male","Female","Male",
              "Female","Male","Female","Male","Male"]
})

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Marks",
    palette="Set2"
)

sns.stripplot(
    data=df,
    x="Gender",
    y="Marks",
    color="black",
    size=6
)

plt.title("Marks Distribution by Gender")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set figure size
plt.figure(figsize=(8, 5))

# Box plot
sns.boxplot(
    data=df,
    x="Gender",
    y="Marks",
    palette="Set2"
)

# Strip plot (overlay)
sns.stripplot(
    data=df,
    x="Gender",
    y="Marks",
    color="black",
    size=6,
    jitter=True
)

# Title and labels
plt.title("Marks Distribution with Individual Students")
plt.xlabel("Gender")
plt.ylabel("Marks")

# Add light horizontal grid
plt.grid(axis="y", alpha=0.3)

# Show the plot
plt.show()
```

## Heatmaps & Correlation Matrix (One of the Most Important Lessons)

```python
import pandas as pd

df = pd.DataFrame({
    "Study_Hours":[10,15,8,12,18,20,7,16,14,11],
    "Marks":[85,92,78,88,95,97,72,90,84,80],
    "Attendance":[90,95,82,88,98,99,80,93,89,85]
})
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr()

plt.figure(figsize=(6,5))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5,linecolor="white")

plt.show()
```

Popular Color maps: "Blues", "Reds", "Greens", "coolwarm", "viridis", "magma", "rocket","YlGnBu"

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Study_Hours":[10,15,8,12,18,20,7,16,14,11],
    "Marks":[85,92,78,88,95,97,72,90,84,80],
    "Attendance":[90,95,82,88,98,99,80,93,89,85]
})

corr = df.corr()

plt.figure(figsize=(7,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=1,
    linecolor="white",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
corr = df.corr()

# Set figure size
plt.figure(figsize=(6, 5))

# Create the heatmap
sns.heatmap(
    corr,
    annot=True,
    cmap="Blues"
)

# Add title
plt.title("Student Correlation Matrix")

# Show the plot
plt.show()
```

## Pair Plot & Joint Plot (Automatic Exploratory Data Analysis)
Welcome to one of Seaborn's most powerful lessons.

Imagine you receive a dataset with 30 columns.

Before building a machine learning model, you need to answer questions like:

Which variables are related?
Which variables have a linear relationship?
Which variables contain outliers?
What is the distribution of each variable?

Instead of making 20–30 different graphs, Seaborn can generate many of them automatically.

That's why pairplot() is one of the most famous functions in Seaborn.

```python
import pandas as pd

df = pd.DataFrame({
    "Study_Hours":[10,15,8,12,18,20,7,16,14,11],
    "Marks":[85,92,78,88,95,97,72,90,84,80],
    "Attendance":[90,95,82,88,98,99,80,93,89,85],
    "Assignments":[8,10,7,9,10,10,6,9,8,8],
    "Gender":[
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Male"
    ]
})
```

```python
# Pairplot compares many variables.
import seaborn as sns

sns.pairplot(df,diag_kind="kde",hue="Gender",palette="Set2")

plt.show()
```

```python
import seaborn as sns

sns.pairplot(df,diag_kind="kde",hue="Gender",palette="Set2",corner=True)

plt.show()
```

```python
import seaborn as sns

sns.pairplot(df,diag_kind="kde",hue="Gender",palette="Set2",corner=True,height=3) # each graph height=3 inches

plt.show()
```

```python
# Joint Plot, Jointplot focuses on only two variables.
sns.jointplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    kind="scatter"
)
```

```python
# Joint Plot, Jointplot focuses on only two variables.
sns.jointplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    kind="kde"
)
```

```python
# Joint Plot, Jointplot focuses on only two variables.
sns.jointplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    kind="hex"
)
```

```python
sns.jointplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    kind="reg",
    height=7,
    color="purple"
)

plt.show()
```

```python
sns.pairplot(df, height=3)

sns.jointplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    height=7
)
```

```python
#Themes, Colors & Professional Visualizations
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme()

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Study_Hours":[10,15,8,12,18,20,7,16,14,11],
    "Marks":[85,92,78,88,95,97,72,90,84,80],
    "Gender":[
        "Male","Female","Male","Female","Male",
        "Female","Male","Female","Male","Male"
    ]
})

# -------------------------
# Theme
# -------------------------
sns.set_theme(
    style="whitegrid",
    palette="Set2",
    context="notebook"
) # Sets the overall aesthetics of the plots, including background style, color palette, and context for sizing elements.

# -------------------------
# Figure
# -------------------------
plt.figure(figsize=(8,5))

# -------------------------
# Plot
# -------------------------
sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    hue="Gender",
    s=120
) 

# -------------------------
# Labels
# -------------------------
plt.title(
    "Relationship Between Study Hours and Marks",
    fontsize=16,
    weight="bold"
)

plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Marks", fontsize=12)

plt.grid(alpha=0.3)

sns.despine() # Removes the top and right spines from the plot for a cleaner look

plt.tight_layout() # Adjusts the padding between and around subplots to prevent overlapping

plt.show()
```
