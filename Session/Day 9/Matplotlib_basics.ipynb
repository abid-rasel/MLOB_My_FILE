## Scatter plot

```python
import matplotlib.pyplot as plt

height = [160,165,170,175,180]
weight = [50,55,63,70,76]

plt.figure(figsize=(8,5))

plt.scatter(
    height,
    weight,
    color="green",
    s=100, # s = Size of the dot, instead of marker size, it is the area of the dot
    alpha=0.7,
    marker="o",
    label="Students"
)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight")
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]

boys = [55, 60, 65, 70, 75]
girls = [50, 54, 60, 66, 72]

# Create a figure
plt.figure(figsize=(8, 5))

# Scatter plot for boys
plt.scatter(
    x,
    boys,
    color="blue",
    marker="o",
    s=100,
    label="Boys"
)

# Scatter plot for girls
plt.scatter(
    x,
    girls,
    color="red",
    marker="s",
    s=100,
    label="Girls"
)

# Customize axes
plt.xlabel("Student Number")
plt.ylabel("Weight (kg)")
plt.title("Weight Comparison of Boys and Girls")

# Add grid
plt.grid(True)

# Show legend
plt.legend()

# Display the graph
plt.show()
```

## Bar plot
#### A bar chart is used to compare different categories.

```python
import matplotlib.pyplot as plt

students = ["Abid", "Sara", "John", "Mina"]
scores = [85, 92, 78, 95]

plt.bar(students, scores)

plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores")

plt.show()
```

```python
# instead of plt.bar, we can use plt.barh to create a horizontal bar chart.
plt.barh(
    students,
    scores
) 

plt.xlabel("Scores")
plt.ylabel("Students")

plt.title("Student Scores")

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Abid", "Sara", "John", "Mina"]

scores = [85, 92, 78, 95]

plt.figure(figsize=(8,5))

plt.bar(
    students,
    scores,
    color=["skyblue", "lightgreen", "salmon", "orange"],
    width=0.6,
    alpha=0.8,
    label="Exam Scores"
)

plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Exam Scores")
plt.grid(axis="y") # instead of plt.grid(True), which creates both horizontal and vertical grid lines.
plt.legend()

plt.show()
```

```python
# Grouped bar chart
import numpy as np
import matplotlib.pyplot as plt

subjects = ["Math", "English", "Science"]

boys = [80,75,90]
girls = [78,82,88]

x = np.arange(len(subjects))

width = 0.35

plt.figure(figsize=(8,5))

plt.bar(
    x - width/2,
    boys,
    width=width,
    label="Boys"
)

plt.bar(
    x + width/2,
    girls,
    width=width,
    label="Girls"
)

plt.xticks(x, subjects)

plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.title("Scores by Subject")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ["CSE", "EEE", "BBA", "Law"]

students = [40,30,20,10]

plt.figure(figsize=(7,7))

plt.pie(
    students,
    labels=departments,
    colors=[
        "skyblue",
        "orange",
        "green",
        "pink"
    ],
    autopct="%1.1f%%",
    explode=[0.1,0,0,0],
    startangle=90,
    shadow=True
)

plt.title("Department Distribution")

plt.show()
```

## Multiple Plots, Multiple Lines, and Legends

```python
# Multiple line plot
import matplotlib.pyplot as plt

months = [1,2,3,4,5]

sales = [20,30,40,45,60]

profit = [5,8,10,15,20]

plt.figure(figsize=(8,5))

plt.plot(
    months,
    sales,
    color="blue",
    label="Sales"
)

plt.plot(
    months,
    profit,
    color="red",
    label="Profit"
)

plt.xlabel("Month")
plt.ylabel("Amount")
plt.title("Sales vs Profit")

plt.legend()

plt.grid(True)

plt.show()
```

```python
# Multiple Scatter Plot
boys = [55,60,65,70]

girls = [52,58,62,68]

x = [1,2,3,4]

plt.scatter(
    x,
    boys,
    color="blue",
    label="Boys"
)

plt.scatter(
    x,
    girls,
    color="red",
    label="Girls"
)

plt.legend()

plt.show()
```

```python
# Mixing plot types (Line + Scatter)
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [2,5,3,7,6]

plt.plot(
    x,
    y,
    color="blue",
    linewidth=2,
    label="Trend"
)

plt.scatter(
    x,
    y,
    color="red",
    s=80,
    label="Actual Data"
)

plt.legend()
#Because label only stores the names. legend() displays them.
plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5]

sales = [20,30,40,45,60]

profit = [5,8,10,15,20]

plt.figure(figsize=(9,5))

plt.plot(
    months,
    sales,
    marker="o",
    linewidth=3,
    label="Sales"
)

plt.plot(
    months,
    profit,
    marker="s",
    linewidth=3,
    label="Profit"
)

plt.xlabel("Month")
plt.ylabel("Amount")
plt.title("Monthly Performance")

plt.grid(True)

plt.legend(
    loc="upper left",
    title="Metrics",
    fontsize=10,
    frameon=True
)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May"]

income = [50, 60, 65, 80, 90]
expense = [35, 40, 45, 55, 60]

# Create the figure
plt.figure(figsize=(10, 6))

# Plot Income
plt.plot(
    months,
    income,
    color="blue",
    marker="o",
    linewidth=2,
    label="Income"
)

# Plot Expense
plt.plot(
    months,
    expense,
    color="red",
    marker="s",
    linewidth=2,
    linestyle="--",
    label="Expense"
)

# Customize the axes
plt.xlabel("Month")
plt.ylabel("Amount ($)")
plt.title("Monthly Income vs Expense")

# Add grid
plt.grid(True)

# Add legend
plt.legend(
    title="Financial Metrics",
    loc="lower right"
)

# Display the graph
plt.show()
```

## Subplots (plt.subplot() and plt.subplots())


```python
# The Old Way: plt.subplot()
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.subplot(2,2,1)
plt.plot([1,2,3],[2,4,6])
plt.title("Graph 1")

plt.subplot(2,2,2)
plt.plot([1,2,3],[6,4,2])
plt.title("Graph 2")

plt.subplot(2,2,3)
plt.bar(["A","B","C"], [3,5,2])
plt.title("Graph 3")

plt.subplot(2,2,4)
plt.scatter([1,2,3],[4,2,5])
plt.title("Graph 4")

plt.tight_layout() # Adjusts the spacing between subplots to prevent overlap, "Please arrange everything neatly."

plt.show()
```

#### fig, ax = plt.subplots() is equivalent to
#### figure = Figure()

#### axes = Axes()
#### Because you're telling a specific Axes to draw the graph.

```python
# Single Subplot Example
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,4))

ax.plot([1,2,3],[2,4,6])

ax.set_title("Sales")

ax.set_xlabel("Month")

ax.set_ylabel("Revenue")

plt.show()
```

```python
# Multiple Subplots
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1,2, figsize=(10,4))

# Left graph
ax[0].plot([1,2,3],[2,4,6])
ax[0].set_title("Sales")

# Right graph
ax[1].bar(["A","B","C"], [5,3,4])
ax[1].set_title("Products")

plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2, figsize=(8,6))

ax[0,0].plot([1,2,3],[2,4,6])
ax[0,0].set_title("Line")

ax[0,1].scatter([1,2,3],[4,2,5])
ax[0,1].set_title("Scatter")

ax[1,0].bar(["A","B","C"], [3,5,2])
ax[1,0].set_title("Bar")

ax[1,1].hist([1,2,2,3,4,4,4,5])
ax[1,1].set_title("Histogram")

plt.tight_layout()
fig.suptitle("Company Performance") # if you want a title for the entire figure

plt.show()
```
